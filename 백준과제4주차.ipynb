```python
T = int(input())
result = []

for _ in range(T):
    A, B = map(int, input().split())
    result.append(str(A + B)) # append는 리스트에 값 넣어주는 코드

print('Wn'.join(result))  #백슬래쉬n으로 하면 열 바꿈 됨
#join 함수가 str 만 쓸 수 있어서 위에서 str로 바꿔서 입력해줌

```
