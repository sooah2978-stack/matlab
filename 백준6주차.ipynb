```python
N = int(input())
nums = list(map(int, input().split())) #한번에 리스트로 입력 받는 법
v = int(input())

count = 0
for i in range(N):
    if nums[i] == v:
        count += 1

print(count)
```

```python
# 첫 번째 줄 입력: 정수 N(개수), X(기준값)을 입력받음
# 예: "10 5"
N, X = map(int, input().split())

# 두 번째 줄 입력: 공백으로 구분된 숫자들을 리스트로 저장
# 예: "1 10 4 9 2 3 8 5 7 6"
nums = list(map(int, input().split()))

# nums 리스트에 있는 값을 하나씩 꺼내면서 반복
for i in nums:

    # 현재 값 i가 기준값 X보다 작은지 확인
    if i < X:

        # 조건을 만족하면 출력
        # end=' ' → 줄바꿈 없이 공백으로 이어서 출력
        print(i, end=' ')
```

```python
# 첫 줄 입력: 바구니 개수 N, 작업 횟수 M 입력받기
# 예: 5 4
N, M = map(int, input().split())

# 바구니 리스트 생성 (처음엔 전부 0 = 공 없음)
# 예: [0, 0, 0, 0, 0]
basket = [0] * N


# 공 넣는 작업을 M번 반복
for _ in range(M):  # _는 그냥 "반복만 할 때" 쓰는 변수

    # i, j, k 입력받기
    # i번 바구니부터 j번 바구니까지 k번 공 넣기
    # 예: 1 2 3
    i, j, k = map(int, input().split())

    # i번부터 j번 바구니까지 반복
    # 파이썬은 0부터 시작하니까 i-1로 맞춰줘야 함
    # range(i-1, j)는 (i-1)부터 (j-1)까지 의미
    for a in range(i - 1, j):

        # 해당 바구니에 k번 공 넣기
        # 기존에 있던 공은 덮어씌워짐
        basket[a] = k


# 최종 바구니 상태 출력
# *basket → 리스트를 공백으로 풀어서 출력
# 예: 1 2 1 1 0
print(*basket)
```
