```python
#문제 1번
T = int(input())

for i in range(T):
  A,B = map(int,input().split())
  C = A + B
  print(f"Case #{i+1}: {C}")
#f"{i+1}"   # 계산도 가능 변수 바로 넣을 수 있음
#{} 👉 변수/계산 넣는 용도
```

```python
#문제 2번
T = int(input())

for i in range(T):
  A,B = map(int,input().split())
  C = A + B
  print(f"Case #{i+1}: {A} + {B} = {C}")
```

```python
#문제 3번

N = int(input())

for i in range(1, N+1): #1부터 시작해야함 i는 0부터니깐..
    print("*" * i)
```

```python
#문제 4번

N = int(input())

for i in range(1, N+1):

    print(" " * (N - i) + "*" * i)
```

```python
#문제 5번
 #for i in range는 값 있을 때만 사용이용...
while True:
    A, B = map(int, input().split())

    if A == 0 and B == 0:
        break

    print(A + B)
```

```python
#문제 6번
while True:
    try:
        A, B = map(int, input().split())
        print(A + B)
    except:
        break
```

```python
while True:
    조건 확인
    if 끝내야 하면:
        break
    실행할 코드

#입력이 더 이상 없으면 input()에서 에러 발생
#그걸 except가 잡아서 반복 종료

while True:
    x = int(input())

    if x == 0:
        break

    print(x)

  #예시 1 (짝수만 출력)

  for i in range(1, 6):
    if i % 2 == 1:
        continue #건너뜀

    print(i)
```
