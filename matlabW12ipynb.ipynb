```python
import sys

def solve():
    input = sys.stdin.read
    data = input().split()

    if not data:
        return

    t = int(data[0])
    results = []

    for i in range(1, t + 1):
        n = int(data[i])

        k = 1
        temp = n
        d = 2

        # sqrt(n)까지 돌며 소인수를 찾습니다.
        while d * d <= temp:
            if temp % d == 0:
                k *= d # 소인수를 결과에 곱해줍니다.
                # 해당 소인수로 나누어 떨어질 때까지 계속 나눕니다 (중복 제거)
                while temp % d == 0:
                    temp //= d
            d += 1

        # 만약 소인수분해 후 남은 수가 1보다 크다면, 그 수 자체가 소수입니다.
        if temp > 1:
            k *= temp

        results.append(str(k))

    print('\n'.join(results))

if __name__ == '__main__':
    solve()
```

```python
import sys
from collections import Counter

def solve():
    input = sys.stdin.read
    data = input().split()

    if not data:
        return

    t = int(data[0])
    idx = 1

    out = []
    for _ in range(t):
        n = int(data[idx])
        idx += 1

        # n * n 개의 사탕 색상 데이터를 가져옵니다.
        candies = []
        for _ in range(n):
            candies.extend(data[idx : idx + n])
            idx += n

        # 각 색상별 사탕의 개수를 셉니다.
        counter = Counter(candies)

        # 가장 많이 존재하는 색상의 개수를 찾습니다.
        max_count = max(counter.values()) if counter else 0

        # 임계값인 n^2 - n 을 계산합니다.
        limit = n * n - n

        # 가장 많은 사탕의 개수가 limit을 초과하면 NO, 아니면 YES
        if max_count > limit:
            out.append("NO")
        else:
            out.append("YES")

    print("\n".join(out))

if __name__ == '__main__':
    solve()
```

```python
import sys

def solve():
    input = sys.stdin.read
    data = input().split()

    if not data:
        return

    t = int(data[0])
    idx = 1

    out = []
    for _ in range(t):
        n = int(data[idx])
        a = [int(x) for x in data[idx+1 : idx+1+n]]
        idx += 1 + n

        # 1. a[0]과 a[n-1]의 값 채우기 규칙 적용
        if a[0] == -1 and a[n-1] == -1:
            a[0] = 0
            a[n-1] = 0
        elif a[0] == -1 and a[n-1] != -1:
            a[0] = a[n-1]
        elif a[0] != -1 and a[n-1] == -1:
            a[n-1] = a[0]

        # 2. 중간에 남은 모든 -1은 사전 순 최소화를 위해 0으로 채움
        for i in range(1, n - 1):
            if a[i] == -1:
                a[i] = 0

        # 3. 최소 절댓값 계산 및 저장
        min_diff = abs(a[n-1] - a[0])

        out.append(str(min_diff))
        out.append(" ".join(map(str, a)))

    print("\n".join(out))

if __name__ == '__main__':
    solve()
```

```python
import sys

def solve():
    input = sys.stdin.read
    data = input().split()

    if not data:
        return

    t = int(data[0])
    idx = 1

    out = []
    for _ in range(t):
        n = int(data[idx])
        k = int(data[idx+1])

        # 배열 a (0-index 기반으로 저장)
        a = [int(x) for x in data[idx+2 : idx+2+n]]

        # 특별한 인덱스 p들 (1-index이므로 0-index로 보정)
        # k=1 이므로 p_indices[0]이 곧 하나뿐인 특별한 인덱스입니다.
        p_indices = [int(x) - 1 for x in data[idx+2+n : idx+2+n+k]]

        idx += 2 + n + k

        # 목표 값 x는 특별한 인덱스에 있는 값
        x = a[p_indices[0]]

        ans = 0

        # 1. p_1의 왼쪽 영역 확인 (p_1에서부터 맨 왼쪽 0까지 역방향 탐색)
        # 값이 달라지는 경계가 생길 때마다 연산 횟수 추가
        current = x
        for i in range(p_indices[0] - 1, -1, -1):
            if a[i] != current:
                ans += 1
                current = a[i]

        # 2. p_1의 오른쪽 영역 확인 (p_1에서부터 맨 오른쪽 n-1까지 정방향 탐색)
        current = x
        for i in range(p_indices[0] + 1, n):
            if a[i] != current:
                ans += 1
                current = a[i]

        out.append(str(ans))

    print("\n".join(out))

if __name__ == '__main__':
    solve()
```
